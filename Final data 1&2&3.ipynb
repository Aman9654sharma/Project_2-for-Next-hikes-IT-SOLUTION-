# 1. FIND BOXPLOT OF "WINDSPEED"

```python
sns.boxplot(x=Final_data['windspeed'])
```

```python
Final_data['windspeed'].describe()
```

# Finding IQR OF Windspeed

```python
percentile25 = Final_data['windspeed'].quantile(0.25)
percentile25 
```

```python
percentile75 = Final_data['windspeed'].quantile(0.75)
percentile75
```

```python
# IQR OF Windspeed = Q3 -Q1
IQR = percentile75 - percentile25 
IQR                     
```

```python
upper_limit = percentile75 + 1.5*IQR
lower_limit = percentile25 - 1.5*IQR

print(upper_limit)
print(lower_limit)
```

```python
print("Mean value of windspeed",Final_data["windspeed"].mean())
print("std value of windspeed",Final_data["windspeed"].std())
print("Min value of windspeed",Final_data["windspeed"].min())
print("Max value of windspeed",Final_data["windspeed"].max())
```

```python
# calculate z-score of windspeed
Final_data["windspeed_zscore"] = Final_data["windspeed"]-Final_data["windspeed"].mean()/Final_data["windspeed"].std()
Final_data["windspeed_zscore"]
```

# FINDING OUTLIER OF WINDSPEED

```python
Final_data[Final_data['windspeed']>upper_limit]
```

```python
Final_data[Final_data['windspeed']<lower_limit]
```

# Caping function used in windspeed

```python
Final_data["windspeed"]=np.where(
    Final_data["windspeed"]>upper_limit,
   upper_limit,np.where(Final_data["windspeed"]<lower_limit,
                         lower_limit,Final_data["windspeed"]))
```

# After Caping no Outlier in "Windspeed"

```python
sns.boxplot(x=Final_data["windspeed"])
```
